<h2>Introdução</h2>

O Teorema de Bayes é um importante teorema da probabilidade utilizado para calcular e atualizar a probabilidade de um evento ou causa $H$ ter ocorrido com base na observação de uma nova evidência $E$, determinando assim a sua probabilidade a posteriori através da relação:

$$P(H \mid E) = \frac{P(E \mid H) \cdot P(H)}{P(E)}$$

Além disso, outro conceito fundamental é a Lei da Probabilidade Total, que permite calcular a probabilidade global desse mesmo evento $E$ ocorrer a partir do somatório das probabilidades condicionadas em diferentes cenários intermediários e mutuamente exclusivos ($A_i$), expressa por:

$$P(E) = \sum_{i} P(E \mid A_i) \cdot P(A_i)$$

Com base nesses fundamentos teóricos, serão resolvidas duas questões que envolvem a determinação e a atualização da probabilidade de uma peça defeituosa ter sido produzida por uma máquina específica, considerando tanto as taxas iniciais de produção quanto as leituras fornecidas por um sensor de verificação.

<h2>Desenvolvimento</h2>

Dada a situação-problema:

Uma fábrica produz peças usando duas máquinas: a Máquina A produz 60% das peças e a Máquina B produz 40%. A Máquina A tem taxa de defeito de 2%, enquanto a Máquina B tem taxa de defeito de 5%.
1. Se uma peça é escolhida aleatoriamente e está com defeito, qual a probabilidade de ter sido produzida pela Máquina B?
2. Quando uma peça defeituosa é encontrada, ela passa por um sensor óptico que tenta identificar a máquina de origem. O sensor acerta a máquina correta em 85% das vezes. A peça defeituosa do exercício 1 foi classificada pelo sensor como vinda da Máquina B. Compute a nova probabilidade de a peça ter vindo da Máquina B, usando a posteriori anterior como priori.

Vamos inicialmente definir os símbolos que usaremos durante a resolução:

$A$: peça foi produzida pela Máquina A

$B$: peça foi produzida pela Máquina B

$D$: peça está com defeito

$S_B$: sensor óptico indicou a peça defeituosa como vinda da Máquina B

<h3>Questão 1</h3>

Como já sabemos que a peça está com defeito e queremos saber a probabilidade de ela ter sido produzida pela Máquina B, estamos em busca do valor de $P(B \mid D)$. Pelo Teorema de Bayes, podemos determiná-la através da fórmula:

$$P(B \mid D) = \frac{P(D \mid B) \cdot P(B)}{P(D)}$$

Desses dados, já temos:

$$P(D \mid B) = 0.05$$

$$P(B) = 0.4$$

Ainda falta, contudo, $P(D)$. Para determiná-lo, vamos usar a lei da probabilidade total:

$$P(D) = P(D \mid B) \cdot P(B) + P(D \mid B^c) \cdot P(B^c) = P(D \mid B) \cdot P(B) + P(D \mid A) \cdot P(A)$$

$$= 0.05 \cdot 0.4 + 0.02 \cdot 0.6 = 0.032$$

Determinado $P(D)$, podemos agora, a partir do Teorema de Bayes, descobrir o valor de $P(B \mid D)$:

$$P(B \mid D) = \frac{P(D \mid B) \cdot P(B)}{P(D)} = \frac{0.05 \cdot 0.4}{0.032} = 0.625 = 62.5\%$$

Assim, conclui-se que a probabilidade de, escolhida aleatoriamente, uma peça ter sido produzida pela Máquina B sabendo que ela é defeituosa é de $62.5\%$.

<h3>Questão 2</h3>

Vamos primeiramente atualizar a probabilidade a priori $P'(B)$ como sendo a posteriori da questão 1, como solicitado pelo enunciado:

$$P'(B) = P(B \mid D) = 0.625$$

o que altera também a probabilidade a priori de $P'(A)$:

$$P'(A) = P(A \mid D) = 1 - 0.625 = 0.375$$

Agora, para responder à questão 2, precisamos calcular $P(B \mid S_B)$. Para isso, vamos novamente usar o Teorema de Bayes:

$$P(B \mid S_B) =\frac{P(S_B \mid B) \cdot P'(B)}{P(S_B)}$$

Desses dados, temos:

$$P'(B) = 0.625$$

$$P(S_B \mid B) = 0.85$$

Ainda falta, contudo, $P(S_B)$. Para determiná-lo, vamos usar a lei da probabilidade total:

$$P(S_B) = P(S_B \mid B) \cdot P'(B) + P(S_B \mid B^c) \cdot P'(B^c) = P(S_B \mid B) \cdot P'(B) + P(S_B \mid A) \cdot P'(A)$$

$$= 0.85 \cdot 0.625 + 0.15 \cdot 0.375 = 0.5875$$

Determinado $P(S_B)$, podemos agora descobrir o valor de $P(B \mid S_B)$:

$$P(B \mid S_B) = \frac{P(S_B \mid B) \cdot P'(B)}{P(S_B)} = \frac{0.85 \cdot 0.625}{0.5875} \approx 0.9043 = 90.43\%$$

Assim, conclui-se que a probabilidade de uma peça ter sido produzida pela Máquina B sabendo que ela é defeituosa e que o sensor a indicou como vinda da Máquina B é de $90.43\%$.

<h2>Conclusão</h2>

Portanto, observa-se que, inicialmente, considerando apenas as taxas de produção e de defeito das máquinas, a probabilidade a priori de uma peça defeituosa ter sido fabricada pela Máquina B era de $62.5\%$. Após a inclusão da leitura do sensor óptico (que possui $85\%$ de precisão), a probabilidade a posteriori foi atualizada para $90.43\%$. Esse aumento expressivo demonstra a eficácia da aplicação conjunta do Teorema de Bayes e da Lei da Probabilidade Total para aumentar a precisão na identificação da origem das falhas.
