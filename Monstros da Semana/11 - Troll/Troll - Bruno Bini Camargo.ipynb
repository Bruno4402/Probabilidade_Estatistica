<h2>Introdução</h2>

O Problema de Monty Hall é um famoso quebra-cabeça de probabilidade baseado no programa de televisão americano *Let's Make a Deal* $[3]$. Na dinâmica do jogo, o participante escolhe uma entre três portas, sabendo que atrás de uma há um carro e nas outras duas há cabras, e o apresentador, que sabe a localização exata do prêmio, abre uma das portas não escolhidas revelando uma cabra e oferece a opção de trocar a escolha inicial pela outra porta fechada. 

Matematicamente, a decisão de trocar de porta eleva a chance de vitória para $\frac{2}{3}$, enquanto manter a escolha original garante apenas $\frac{1}{3}$ de probabilidade de ganhar. Essa vantagem ocorre porque a chance inicial de escolher uma cabra é de $\frac{2}{3}$, de modo que a atitude do apresentador em eliminar uma cabra concentra toda essa probabilidade acumulada de $\frac{2}{3}$ na única porta restante.

Apesar de sua fundamentação matemática, essa conclusão gerou grande controvérsia quando foi popularizada por Marilyn vos Savant em $1990$, sendo questionada por diversos leitores e até por matemáticos renomados. Nesse contexto, simulações computacionais tornaram-se uma ferramenta importante para demonstrar empiricamente os resultados previstos pela teoria. Assim, nesta atividade, será desenvolvida uma simulação em Python do problema de Monty Hall com o objetivo de verificar experimentalmente se a estratégia de trocar de porta realmente apresenta uma taxa de sucesso próxima de $\frac{2}{3}$.

<h2>Análise por simulação</h2>

Para implementar a simulação, inicialmente será criada uma função capaz de reproduzir uma única partida do problema de Monty Hall. Nessa função, serão definidas aleatoriamente a localização do carro e a escolha inicial do participante. Em seguida, o apresentador abrirá uma porta contendo uma cabra e o participante adotará a estratégia de trocar sua escolha para a única porta fechada restante. Ao final, será verificado se a troca resultou na obtenção do prêmio.

```python
import random

def monty_hall():
    """
    Função que simula uma vez o problema de Monty Hall
    """

    porta_carro = random.randint(1, 3)
    escolha_jogador = random.randint(1, 3)
    portas_abriveis = []

    for porta in [1, 2, 3]:
        if porta != escolha_jogador and porta != porta_carro:
            portas_abriveis.append(porta)

    porta_aberta = random.choice(portas_abriveis)

    for porta in [1, 2, 3]:
        if porta != escolha_jogador and porta != porta_aberta:
            escolha_jogador = porta
            break

    return escolha_jogador == porta_carro
```

Com a simulação de uma partida concluída, basta repetir o procedimento um grande número de vezes (nesse caso, será realizado $100$ mil vezes) e contabilizar quantas vitórias foram obtidas. Dessa forma, será possível estimar experimentalmente a taxa de sucesso da estratégia de troca e compará-la com o valor teórico esperado de $\frac{2}{3}$.

```python
N = 100000

vitorias = 0

for _ in range(N):
    if monty_hall():
        vitorias += 1

taxa_vitoria = vitorias / N

print(f"Vitórias: {vitorias}")
print(f"Taxa de vitória: {taxa_vitoria:.4f}")
```

A simulação produziu uma taxa de sucesso próxima de $66.67\%$, valor compatível com a probabilidade teórica de $\frac{2}{3}$. Pequenas variações entre execuções são esperadas devido ao caráter aleatório do experimento, mas tendem a diminuir conforme o número de simulações aumenta. 

Dessa forma, os resultados observados corroboram a solução matemática do problema de Monty Hall, indicando que a estratégia de trocar de porta é, de fato, a alternativa mais vantajosa para maximizar as chances de vitória.

<h2>Análise complementar por teste de hipótese</h2>

Embora a comparação da taxa de vitórias observada com o valor teórico já seja suficiente para validar experimentalmente o problema de Monty Hall, também é possível utilizar um teste de hipótese para verificar se a proporção de vitórias obtida na simulação é compatível com a probabilidade teórica de $\frac{2}{3}$.

Considere $p$ como a probabilidade de vitória ao adotar a estratégia de trocar de porta.
Definindo as hipóteses $H_0$ e $H_1$, tem-se:

- $H_0: p = \frac{2}{3}$
- $H_1: p \neq \frac{2}{3}$

A hipótese nula afirma que a probabilidade de vitória observada na simulação é igual à probabilidade teórica do problema de Monty Hall, enquanto a hipótese alternativa afirma que existe diferença entre esses valores.

Como o resultado de cada execução da simulação pode ser classificado apenas como vitória ou derrota, a variável analisada é binária. Além disso, o interesse está em verificar se a proporção de vitórias observada é compatível com uma proporção teórica conhecida, igual a $\frac{2}{3}$.

Dessa forma, o teste estatístico mais adequado é o Teste Z para uma proporção $[2]$, que permite comparar a proporção observada em uma amostra com um valor hipotético estabelecido na hipótese nula.

Antes da aplicação do teste, é necessário verificar se seus requisitos são atendidos.

O primeiro requisito exige que a variável de interesse possua apenas dois resultados possíveis.
No contexto desta simulação, cada execução pode resultar em:

- Vitória: o participante vence após trocar de porta
- Fracasso: o participante não vence após trocar de porta

Portanto, a variável analisada é binária, satisfazendo esse requisito.

----

O segundo requisito exige que os dados sejam obtidos de forma aleatória.
Na simulação implementada, tanto a localização do carro quanto a escolha inicial do participante são determinadas por sorteios aleatórios realizados pela biblioteca `random` do Python.
Assim, os dados produzidos podem ser considerados aleatórios.

----

O terceiro requisito estabelece que as observações devem ser independentes entre si.
Cada execução da função `monty_hall()` gera uma nova partida sem utilizar informações provenientes das execuções anteriores. Como consequência, o resultado de uma simulação não influencia o resultado das demais.

No Teste Z para uma proporção, a condição dos $10\%$ é frequentemente utilizada em situações de amostragem sem reposição em populações finitas. Essa condição estabelece que o tamanho da amostra ($n$) não deve exceder $10\%$ do tamanho da população ($N$), isto é, $n \leq 0.10N$. O objetivo dessa condição é garantir que as observações possam ser consideradas aproximadamente independentes.

Entretanto, neste trabalho os dados não foram coletados de uma população existente, mas sim gerados por uma simulação computacional independente, de modo que essa condição não se aplica. Portanto, o pressuposto de independência pode ser considerado atendido.

-----

O último requisito exige que o número esperado de vitórias e fracassos, considerando a hipótese nula, seja suficientemente grande.
Como $H_0 : p = \frac{2}{3}$ e foram realizadas $n = 100$ mil simulações, tem-se:

```python
p0 = 2/3
n = 100000

print(f"Vitórias = {n*p0: .2f}")
print(f"Fracassos = {n*(1-p0): .2f}")
```

Como ambos os valores são muito superiores a $10$, a condição de sucesso e fracasso é amplamente satisfeita.

Dessa forma, **todos** os requisitos relevantes para a aplicação do Teste Z para uma proporção podem ser considerados atendidos. 

Antes de realizar o teste, é interessante enunciar uma função para concluir o teste de hipótese a partir do $p$-valor observado. A função foi retirada do material de aula "LMA-203 A.4 - Usando testes de hipóteses prontos usuais", do Prof. Dr. Daniel Cassar, para a disciplina de Probabilidade e Estatística, da Ilum Escola de Ciência $[1]$.

```python
def conclusao(pvalor, significancia):

    print(f"p-valor = {pvalor: .6f}")

    print()

    if pvalor < significancia:
        print("Conclusão do teste de hipótese: \033[31mdevemos rejeitar H0\033[0m")
    else:
        print("Conclusão do teste de hipótese: \x1b[32mnão devemos rejeitar H0\x1b[0m")
```

Em seguida, o teste pode ser realizado utilizando a proporção observada na simulação:

```python
from statsmodels.stats.proportion import proportions_ztest

sucessos = vitorias
n = N

z, p_valor = proportions_ztest(
    count=sucessos,
    nobs=n,
    value=2/3
)

print(f"Estatística Z = {z:.6f}")
conclusao(p_valor, 0.05)
```

Foi obtido um $p$-valor superior ao nível de significância adotado ($\alpha = 0.05$). Dessa forma, não há evidências suficientes para rejeitar $H_0$, indicando que a taxa de vitórias observada é compatível com a probabilidade teórica do problema de Monty Hall. Esse resultado reforça a conclusão obtida pela simulação e está de acordo com a afirmação de Marilyn vos Savant de que a estratégia de trocar de porta possui probabilidade de vitória igual a $\frac{2}{3}$.

<h2>Conclusão</h2>

A simulação computacional realizada reproduziu com sucesso a dinâmica do problema de Monty Hall e permitiu estimar experimentalmente a probabilidade de vitória associada à estratégia de trocar de porta. Com $100$ mil execuções, foi obtida uma taxa de sucesso próxima a $66.67\%$, valor muito próximo da probabilidade teórica de $\frac{2}{3}$.

Além disso, o Teste Z para uma proporção resultou em um $p$-valor acima do valor de significância de $\alpha = 0.05$, não fornecendo evidências para rejeitar a hipótese de que a probabilidade de vitória seja igual a $\frac{2}{3}$. Assim, os resultados obtidos são compatíveis com a teoria probabilística do problema.

Dessa forma, tanto a simulação quanto a análise estatística mostram que a solução apresentada por Marilyn vos Savant estava correta: trocar de porta aumenta as chances de ganhar o prêmio, enquanto permanecer com a escolha inicial oferece uma probabilidade de sucesso menor.
